```python
import pandas as pd
from pathlib import Path

# Locate the dataset relative to the notebook
data_path = Path("../data/ai4i2020.csv")

# Load the dataset
df = pd.read_csv(data_path)

# Display basic information
print("Dataset loaded successfully.")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")

display(df.head())
```

```python
# Inspect the dataset structure

print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
display(df.isnull().sum())

print("\nStatistical summary:")
display(df.describe())
```

```python
# Check data quality
print("Dataset shape:", df.shape)
print("\nMissing values per column:")
display(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

# Count machine failures
print("\nMachine failure counts:")
display(df["Machine failure"].value_counts())

# Count each failure mode
failure_columns = ["TWF", "HDF", "PWF", "OSF", "RNF"]

print("\nFailure mode counts:")
display(df[failure_columns].sum().sort_values(ascending=False))
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Select operating-condition variables
feature_columns = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

# Keep only the selected features
X = df[feature_columns].copy()

# Encode product type as numeric indicator columns
X = pd.get_dummies(X, columns=["Type"], dtype=int)

# Standardize all clustering features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Original feature columns:", feature_columns)
print("Encoded feature columns:", X.columns.tolist())
print("Prepared data shape:", X_scaled.shape)
print("All values finite:", pd.notna(X_scaled).all())

display(pd.DataFrame(X_scaled, columns=X.columns).head())
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Fit K-Means with 3 clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

# Evaluate clustering quality
silhouette = silhouette_score(X_scaled, cluster_labels)
davies_bouldin = davies_bouldin_score(X_scaled, cluster_labels)
calinski_harabasz = calinski_harabasz_score(X_scaled, cluster_labels)

print("K-Means clustering results")
print("--------------------------")
print("Number of clusters:", len(set(cluster_labels)))
print(f"Silhouette score: {silhouette:.4f}")
print(f"Davies-Bouldin index: {davies_bouldin:.4f}")
print(f"Calinski-Harabasz score: {calinski_harabasz:.2f}")

print("\nRecords per cluster:")
print(pd.Series(cluster_labels).value_counts().sort_index())
```

```python
# Attach cluster assignments to a copy of the original data
results = df.copy()
results["Cluster"] = cluster_labels

# Compare operating conditions across clusters
profile_columns = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

print("Mean operating conditions by cluster:")
display(
    results.groupby("Cluster")[profile_columns]
    .mean()
    .round(2)
)

print("\nMachine failure rate by cluster (%):")
display(
    (results.groupby("Cluster")["Machine failure"].mean() * 100)
    .round(2)
)

print("\nFailure counts by cluster:")
display(
    results.groupby("Cluster")[
        ["Machine failure", "TWF", "HDF", "PWF", "OSF", "RNF"]
    ].sum().astype(int)
)
```

```python
# Product type distribution within each cluster
print("Product type counts by cluster:")
display(pd.crosstab(results["Cluster"], df["Type"]))

print("\nProduct type percentages within each cluster:")
display(
    pd.crosstab(
        results["Cluster"],
        df["Type"],
        normalize="index"
    ).mul(100).round(2)
)

# Compare variability, not just averages
print("\nOperating-condition variability by cluster:")
display(
    results.groupby("Cluster")[profile_columns]
    .std()
    .round(2)
)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Use only continuous operating-condition variables
continuous_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

X_continuous = df[continuous_features].copy()

# Standardize the continuous variables
scaler_continuous = StandardScaler()
X_continuous_scaled = scaler_continuous.fit_transform(X_continuous)

# Fit K-Means with 3 clusters
kmeans_continuous = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters_continuous = kmeans_continuous.fit_predict(
    X_continuous_scaled
)

# Evaluate clustering
print("K-Means without product type")
print("----------------------------")
print(f"Silhouette score: {silhouette_score(X_continuous_scaled, clusters_continuous):.4f}")
print(f"Davies-Bouldin index: {davies_bouldin_score(X_continuous_scaled, clusters_continuous):.4f}")
print(f"Calinski-Harabasz score: {calinski_harabasz_score(X_continuous_scaled, clusters_continuous):.2f}")

# Attach the new cluster assignments
results_continuous = df.copy()
results_continuous["Cluster"] = clusters_continuous

print("\nRecords per cluster:")
display(results_continuous["Cluster"].value_counts().sort_index())

print("\nMean operating conditions by cluster:")
display(
    results_continuous.groupby("Cluster")[continuous_features]
    .mean()
    .round(2)
)
```

```python
# Attach revised cluster assignments
results_continuous = df.copy()
results_continuous["Cluster"] = clusters_continuous

# Observed failure rates by cluster
failure_summary = results_continuous.groupby("Cluster").agg(
    Records=("Machine failure", "size"),
    Failures=("Machine failure", "sum"),
    Failure_Rate=("Machine failure", "mean")
)

failure_summary["Failure_Rate_Percent"] = (
    failure_summary["Failure_Rate"] * 100
)

print("Observed machine failure rates by cluster:")
display(failure_summary.round(3))

# Compare individual failure modes
failure_columns = ["TWF", "HDF", "PWF", "OSF", "RNF"]

print("\nFailure-mode counts by cluster:")
display(
    results_continuous.groupby("Cluster")[failure_columns]
    .sum()
    .astype(int)
)

print("\nFailure-mode rates by cluster (%):")
display(
    results_continuous.groupby("Cluster")[failure_columns]
    .mean()
    .mul(100)
    .round(2)
)
```

```python
import pandas as pd
from scipy.stats import chi2_contingency
import numpy as np

# Build a contingency table: cluster versus machine failure
contingency = pd.crosstab(
    results_continuous["Cluster"],
    results_continuous["Machine failure"]
)

print("Contingency table (rows = cluster, columns = failure label):")
display(contingency)

# Chi-square test of independence
chi2, p_value, dof, expected = chi2_contingency(contingency)

# Cramer's V effect size
n = contingency.to_numpy().sum()
min_dimension = min(contingency.shape[0] - 1, contingency.shape[1] - 1)
cramers_v = np.sqrt(chi2 / (n * min_dimension))

print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.6g}")
print(f"Cramer's V: {cramers_v:.4f}")

print("\nExpected counts under independence:")
display(
    pd.DataFrame(
        expected,
        index=contingency.index,
        columns=contingency.columns
    ).round(2)
)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
import numpy as np

# Fit K-Means 10 times using different random seeds
seeds = range(10)
labels_by_seed = []

for seed in seeds:
    model = KMeans(
        n_clusters=3,
        random_state=seed,
        n_init=10
    )
    labels = model.fit_predict(X_continuous_scaled)
    labels_by_seed.append(labels)

# Compare all pairs of clustering results
ari_scores = []

for i in range(len(labels_by_seed)):
    for j in range(i + 1, len(labels_by_seed)):
        ari_scores.append(
            adjusted_rand_score(
                labels_by_seed[i],
                labels_by_seed[j]
            )
        )

print(f"Number of runs: {len(seeds)}")
print(f"Pairwise comparisons: {len(ari_scores)}")
print(f"Mean ARI: {np.mean(ari_scores):.4f}")
print(f"Minimum ARI: {np.min(ari_scores):.4f}")
print(f"Maximum ARI: {np.max(ari_scores):.4f}")
```

```python
# TEST 4: Logistic regression adjusted for product type

import pandas as pd
import statsmodels.formula.api as smf
import numpy as np

# Prepare the analysis dataset using standardized clustering assignments
analysis_df = df.copy()
analysis_df["Cluster"] = clusters_continuous

# Treat cluster and product type as categorical variables
analysis_df["Cluster"] = analysis_df["Cluster"].astype("category")
analysis_df["Type"] = analysis_df["Type"].astype("category")

# Set Cluster 1 and Type L as reference categories
analysis_df["Cluster"] = analysis_df["Cluster"].cat.reorder_categories(
    [1, 0, 2], ordered=False
)
analysis_df["Type"] = analysis_df["Type"].cat.reorder_categories(
    ["L", "H", "M"], ordered=False
)

# Fit logistic regression
model = smf.logit(
    'Q("Machine failure") ~ C(Cluster) + C(Type)',
    data=analysis_df
).fit(disp=False)

# Convert coefficients into adjusted odds ratios and confidence intervals
conf = model.conf_int()

results = pd.DataFrame({
    "Coefficient": model.params,
    "Odds_Ratio": np.exp(model.params),
    "CI_Lower": np.exp(conf[0]),
    "CI_Upper": np.exp(conf[1]),
    "P_Value": model.pvalues
})

print("Logistic regression adjusted for product type")
print(f"Observations: {int(model.nobs)}")
print(f"Model converged: {model.mle_retvals['converged']}")

display(results.round(4))
```

```python

# FINAL TEST: Cluster × Product Type Interaction
# Research: Predictive Maintenance Through Unsupervised Clustering

import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from scipy.stats import chi2

# 1. Prepare the analysis data
interaction_df = df.copy()
interaction_df["Cluster"] = clusters_continuous

interaction_df["Cluster"] = pd.Categorical(
    interaction_df["Cluster"],
    categories=[1, 0, 2]
)

interaction_df["Type"] = pd.Categorical(
    interaction_df["Type"],
    categories=["L", "H", "M"]
)

# 2. Fit the model without an interaction
model_base = smf.logit(
    'Q("Machine failure") ~ C(Cluster) + C(Type)',
    data=interaction_df
).fit(disp=False)

# 3. Fit the model with the interaction
model_interaction = smf.logit(
    'Q("Machine failure") ~ C(Cluster) * C(Type)',
    data=interaction_df
).fit(disp=False)

# 4. Compare the models using a likelihood-ratio test
lr_statistic = 2 * (
    model_interaction.llf - model_base.llf
)
df_difference = int(
    model_interaction.df_model - model_base.df_model
)
lr_p_value = chi2.sf(lr_statistic, df_difference)

print("FINAL TEST: Cluster × Product Type Interaction")
print("-" * 55)
print(f"Observations: {int(model_interaction.nobs)}")
print(f"Interaction model converged: "
      f"{model_interaction.mle_retvals['converged']}")
print(f"Likelihood-ratio statistic: {lr_statistic:.4f}")
print(f"Degrees of freedom: {df_difference}")
print(f"Interaction p-value: {lr_p_value:.6g}")

# 5. Display interaction coefficients and confidence intervals
conf = model_interaction.conf_int()

interaction_results = pd.DataFrame({
    "Coefficient": model_interaction.params,
    "Odds_Ratio": np.exp(model_interaction.params),
    "CI_Lower": np.exp(conf[0]),
    "CI_Upper": np.exp(conf[1]),
    "P_Value": model_interaction.pvalues
})

print("\nInteraction model coefficients:")
display(interaction_results.round(4))

# 6. Report the overall interaction test
if lr_p_value < 0.05:
    print(
        "\nConclusion: The data provide evidence that the "
        "cluster–failure association differs by product type."
    )
else:
    print(
        "\nConclusion: The data do not provide sufficient "
        "evidence that the cluster–failure association differs "
        "by product type."
    )
```
